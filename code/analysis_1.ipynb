```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import scipy.stats as stats
from scipy.stats import fisher_exact
```

```python
# Data

labels = ['Women', 'Men']
homicides_tot = 314
homicides = [111, 203]
homicides_family = [96, 55]
homicides_family_partner = [59, 10]

# Tabella di contingenza: frequenze osservate
observed = np.array([[55, 148],  # Uomini uccisi in famiglia vs altri contesti
                     [96, 15]])  # Donne uccise in famiglia vs altri contesti
```

```python
# IPOTESI NULLA (H0): non è presente una disparità significativa tra le percentuali di donne e uomini uccisi in ambito familiare
# (La percentuale di donne uccise in contesti familiari è uguale a quella degli uomini.)
# (Se la percentuale di donne uccise in contesti familiari differisce da quella degli uomini, tale differenza è esclusivamente dovuta al caso.)
# IPOTESI ALTERNATIVA (H1): è presente una disparità significativa
# (La percentuale di donne uccise in contesti familiari è significativamente diversa da quella degli uomini, la differenza non è dovuta al caso.)
```

```python
# Chi-quadrato Test: 
# confronto i due gruppi (uomini e donne) per verificare se esiste una differenza significativa nelle distribuzioni delle variabili
# Come funziona il test:
# confronta le frequenze osservate con quelle attese (se non ci fosse alcuna differenza significativa tra uomini e donne)
# default, test del Chi-quadrato di Pearson

# alpha, livello di significatività: probabilità che lo studio rigetti l'ipotesi nulla quando questa è vera
alpha = 0.001 # (0.1 %)

# p_value: probabilità di sbagliarsi nel rigettare l'ipotesi nulla
# dof, grado di libertà: n-1, con n pari al num di variabili analizzate, 2-1=1
chi2, p_value, dof, expected = stats.chi2_contingency(observed)

# Risultati
print("Chi2 statistic:", chi2)
print("p-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)

# Interpretazione del p-value
if p_value < alpha:
    # Rifiuto l'ipotesi nulla: concludo che è presente una differenza significativa
    # il risultato è significativo, quindi la differenza non è dovuta al caso ma è reale (dovuta a qualche fattore)
    print("\nLa differenza tra le percentuali di omicidi in contesti familiari per donne e uomini è statisticamente significativa.")
else:
    # Non c'è abbastanza evidenza per rifiutare l'ipotesi nulla
    print("\nNon c'è una differenza significativa tra le percentuali di omicidi in contesti familiari per donne e uomini.")
```

```python
observed = np.array([
    [96, 15],   # DONNE
    [55, 148]   # UOMINI
])

# Calcolo Odds Ratio e p-value (usando il test esatto di Fisher)
odds_ratio, p_value = fisher_exact(observed)

print(f'Odds Ratio: {odds_ratio}')
print(f'P-value (Fisher test): {p_value}')
```
