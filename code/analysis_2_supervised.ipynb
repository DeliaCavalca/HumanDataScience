```python
from sklearn.preprocessing import OneHotEncoder, LabelEncoder

# LabelEncoder: encode target label with value between 0 and n_classes-1
def labelEncoding(y_train, y_test):

    label_encoder = LabelEncoder()
    y_train_enc = label_encoder.fit_transform(y_train)
    y_test_enc = label_encoder.transform(y_test)

    # label mapping
    class_mapping = label_encoder.classes_
    
    print("\nMapping delle classi in y:")
    for index, label in enumerate(class_mapping):
        print(f"{label} -> {index}")

    return y_train_enc, y_test_enc

# Label Encoding for Feature Target
y_train, y_test = labelEncoding(y_train, y_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, accuracy_score, f1_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn import preprocessing

models = [LogisticRegression(random_state=seed, class_weight="balanced"), SVC(random_state=seed, class_weight='balanced'), RandomForestClassifier(random_state=seed, class_weight='balanced')]
```

```python
# Normalization

def best_normalization(X, y, numerical_features, categorical_features, model, cv=5):
    scores = {}

    # Define the normalization techniques
    normalization_methods = {
        'L1': preprocessing.Normalizer(norm='l1'),
        'L2': preprocessing.Normalizer(norm='l2'),
        'Standardization': preprocessing.StandardScaler(),
        'Min-Max Scaling': preprocessing.MinMaxScaler()
    }

    print(f"\nEvaluating normalization methods on model: {model}")

    scoring = {
        'accuracy': make_scorer(accuracy_score),
        'f1_weighted': make_scorer(f1_score, average='weighted')
    }

    for method_name, normalizer in normalization_methods.items():

        # Preprocessing Step: Feature Encoding and Normalization
        preprocessor = ColumnTransformer(
            transformers=[
                # Encoding for Categorical Features
                ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features),
                # Normalization for Numerical Features
                ('num', normalizer, numerical_features)
            ],
        )

        # Pipeline Definition: base
        pipeline = Pipeline(steps=[
            ('preprocessor', preprocessor), # encoding and normalization
            ('model', model) # modeling
        ])

        # Execute cross-validation
        cv_results = cross_validate(pipeline, X, y, cv=StratifiedKFold(cv, random_state=seed, shuffle=True), scoring=scoring, n_jobs=1)

        # Compute metrics
        acc = cv_results['test_accuracy'].mean()
        f1 = cv_results['test_f1_weighted'].mean()

        scores[method_name] = {'accuracy': acc, 'f1_score': f1}
        print(f"{method_name} - Accuracy (CV): {acc}, F1-score (CV): {f1}")

    return scores

for model in models:
    best_normalization(X_train, y_train, numerical_features, categorical_features, model)
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV, StratifiedKFold

def hyperparameters_tuning(X_train, y_train, pipeline, param_grid, search_type='grid', n_iter=50, scoring='f1_weighted', seed=42):
    
    # Hyperparameters Tuning

    # Stratified K Fold cross-validation
    skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=seed)

    # Define the search type
    if search_type == 'grid':
        search = GridSearchCV(pipeline, param_grid=param_grid, cv=skf, verbose=0, scoring=scoring)
    elif search_type == 'random':
        search = RandomizedSearchCV(pipeline, param_distributions=param_grid, n_iter=n_iter, cv=skf, random_state=seed, n_jobs=-1, verbose=2, scoring=scoring)
    else:
        raise ValueError("search_type must be 'grid' or 'random'")
    
    # Fit on Train Set
    search.fit(X_train, y_train)

    best_model = search.best_estimator_
    print("Best hyperparameters:", search.best_params_)
    print("Best cross-validated accuracy:", search.best_score_)
    
    # Return the Model
    return best_model

from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score, precision_score, recall_score, confusion_matrix

def evaluate_perf(X, y, model):
    
    ypred = model.predict(X)
   
    acc = accuracy_score(y, ypred)
    bacc = balanced_accuracy_score(y, ypred)
    f1 = f1_score(y, ypred, average="weighted")
    precision = precision_score(y, ypred, average="weighted", zero_division=0)
    recall = recall_score(y, ypred, average="weighted", zero_division=0)

    
    # Specificity
    cm = confusion_matrix(y, ypred)
    tn = cm.sum() - (cm.sum(axis=0) + cm.sum(axis=1) - cm.diagonal())
    fp = cm.sum(axis=0) - cm.diagonal()
    specificity = (tn / (tn + fp + 1e-6)).mean()  # No division by 0
    
    perf = {
        "accuracy": acc,
        "balanced_accuracy": bacc,
        "f1_score": f1,
        "precision": precision,
        "recall": recall,
        "specificity": specificity
    }
    
    return perf, cm
```

```python
# Permutation Importance
from sklearn.inspection import permutation_importance
# Model-agnostic method for calculating feature importance.
# Model-agnostic means that the method can be used with any model.

# metodo per valutare l'importanza delle features dopo che il modello è stato addestrato
# controlla quanto peggiora la performance del modello quando rimescoli casualmente i valori di una variabile

def perm_importance(model, X_test, y_test, selected_features, seed=seed): 

    # compute the permutation importance
    result = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=seed)

    # sort features based on the importance
    perm_sorted_idx = result.importances_mean.argsort()[-10:]
    sorted_features = np.array(selected_features)[perm_sorted_idx]

    tick_labels_parameter_name = (
        "tick_labels"
    )
    tick_labels_dict = {tick_labels_parameter_name: sorted_features}

    # plot
    fig, ax = plt.subplots(figsize=(12, 10))
    ax.boxplot(result.importances[perm_sorted_idx].T, vert=False, **tick_labels_dict)

    ax.axvline(x=0, color="black", linestyle="--")
    ax.set_xlabel("Decrease in accuracy score", fontsize=23, labelpad=20)
    ax.set_ylabel('')
    ax.set_title('Permutation Importance with St. Dev.', fontsize=26, pad=20)
    plt.xticks(fontsize=22)
    plt.yticks(fontsize=22)
    plt.legend().set_visible(False)

    plt.tight_layout()
    plt.show()


# Coefficient Importance
def coef_importance(coefficients, selected_features):
    avg_importance = np.mean(np.abs(coefficients), axis=0)
    feature_importance = pd.DataFrame({'Feature': selected_features, 'Importance': avg_importance})
    feature_importance = feature_importance.sort_values('Importance', ascending=True)
    
    ax = feature_importance.plot(x='Feature', y='Importance', kind='barh', figsize=(10, 6))
    ax.set_xlabel('Coefficient Importance')
    ax.set_title('Coefficient Importance')
```

```python

def plot_model_performance(train_scores, test_scores):
   
    models = ['LR', 'SVM', 'RF']
    train_accuracies = [train_scores[model] for model in models]
    test_accuracies = [test_scores[model] for model in models]

    x = np.arange(len(models))  # posizioni per i gruppi di barre
    width = 0.35  # larghezza di ogni barra

    fig, ax = plt.subplots()
    bars1 = ax.bar(x - width/2, train_accuracies, width, label='Train', color='skyblue')
    bars2 = ax.bar(x + width/2, test_accuracies, width, label='Test', color='salmon')

    # Etichette e titolo
    ax.set_ylabel('Accuratezza')
    ax.set_title('Confronto accuratezza modelli su Train e Test set')
    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.legend()

    # Mostra i valori sopra le barre
    def autolabel(bars):
        for bar in bars:
            height = bar.get_height()
            ax.annotate(f'{height:.2f}',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 3),  # 3 punti di offset verticale
                        textcoords="offset points",
                        ha='center', va='bottom')

    autolabel(bars1)
    autolabel(bars2)

    plt.ylim(0, 1.05)
    plt.tight_layout()
    plt.show()
```

```python
train_scores = {'LR': 0., 'SVM': 0., 'RF': 0.}
test_scores = {'LR': 0., 'SVM': 0., 'RF': 0.}

plot_model_performance(train_scores, test_scores)
```
