```python
import math

# Analisi correlazione fra variabili categoriche

# Tabella di contingenza tra tipo di relazione e causa di morte
contingency_table = pd.crosstab(df['Relazione con la persona uccisa'], df['Causa di morte'])

# Test del chi quadrato
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

# Calcolo di Cramér’s V
n = contingency_table.sum().sum()
phi2 = chi2_stat / n
r, k = contingency_table.shape
cramers_v = math.sqrt(phi2 / min(k - 1, r - 1))

# Stampa dei risultati
print(f"Chi-squared Statistic: {chi2_stat:.4f}")
print(f"P-Value: {p_value:.4g}")
print(f"Cramér’s V: {cramers_v:.4f}")

# Calcolo dei residui standardizzati
observed = contingency_table.values
residuals = (observed - expected) / np.sqrt(expected)

# Metti i residui in un DataFrame leggibile
residuals_df = pd.DataFrame(residuals,
                            index=contingency_table.index,
                            columns=contingency_table.columns)

print("\nResidui standardizzati:")
print(residuals_df.round(2))
```
