```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import scipy.stats as stats

seed = 42
np.random.seed(seed)
```

```python
# Data

FILEPATH = '.\\femminicidi ITA 2024.xlsx'
df_2024 = pd.read_excel(FILEPATH, header=0)
FILEPATH = '.\\femminicidi ITA 2023.xlsx'
df_2023 = pd.read_excel(FILEPATH, header=0)
FILEPATH = '.\\femminicidi ITA 2022.xlsx'
df_2022 = pd.read_excel(FILEPATH, header=0)
FILEPATH = '.\\femminicidi ITA 2021.xlsx'
df_2021 = pd.read_excel(FILEPATH, header=0)
FILEPATH = '.\\femminicidi ITA 2020.xlsx'
df_2020 = pd.read_excel(FILEPATH, header=0)

print("Righe: " + str(df_2024.shape[0]) + " - Colonne: " + str(df_2024.shape[1]))
print("Righe: " + str(df_2023.shape[0]) + " - Colonne: " + str(df_2023.shape[1]))
print("Righe: " + str(df_2022.shape[0]) + " - Colonne: " + str(df_2022.shape[1]))
print("Righe: " + str(df_2021.shape[0]) + " - Colonne: " + str(df_2021.shape[1]))
print("Righe: " + str(df_2020.shape[0]) + " - Colonne: " + str(df_2020.shape[1]))
```

```python
# get only 2024 data
df = df_2024

df = df[df['Tipo'] == 'omicidio']
df = df[df['Relazione con la persona uccisa'] != '-']

print("Righe: " + str(df.shape[0]) + " - Colonne: " + str(df.shape[1]))
```

```python
# Analysis 2024

observed = df['Relazione con la persona uccisa'].value_counts()
print(observed)

num_categories = len(observed)

total_observations = observed.sum()

expected = [total_observations / num_categories] * num_categories 

chi_squared_stat, p_value = stats.chisquare(observed, expected)

print(f"\nChi-Squared Statistic: {chi_squared_stat}")
print(f"P-Value: {p_value}")
```
