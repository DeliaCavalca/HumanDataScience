Unsupervised Learning

```python
from sklearn.cluster import KMeans
from yellowbrick.cluster import KElbowVisualizer, SilhouetteVisualizer
import umap

""" Dimensionality Reduction and Data Visualization """

def umap_reduction(X, labels, title, n_components=2, n_neighbors=15, min_dist=0.1, centroids=None, palette='pastel', color_label=None, x_col=None):
    plt.figure(figsize=(12,12))
    
    # Dimensionality Reduction
    reducer = umap.UMAP(random_state=seed, n_components=n_components, n_neighbors=n_neighbors, min_dist=min_dist)
    X_umap = reducer.fit_transform(X)

    unique_labels = np.unique(labels)
    colors = sns.color_palette(palette, len(unique_labels))
    color = sns.color_palette(palette, 1)

    if n_components == 1:
        data = pd.DataFrame(X_umap, columns=['Component 1'])
        data[color_label] = labels
        if x_col is not None:
            sns.swarmplot(data=data, x=x_col, y='Component 1', hue=color_label, palette=palette, alpha=0.7, size=3)
        else:
            sns.swarmplot(data=data, y='Component 1', hue=color_label, palette=palette, alpha=0.7, size=3)
        
        plt.xlabel("1D Projection")
            
    else:
        for i, label in enumerate(unique_labels):
            cluster_points = X_umap[labels == label]
            plt.scatter(cluster_points[:, 0], cluster_points[:, 1], s=30, color=colors[i], label=f"{label}")

            #plt.scatter(cluster_points[:, 0], cluster_points[:, 1], s=30, color=color, label=f"{label}")

    if centroids is not None:
        centroids_umap = reducer.transform(centroids) 
        plt.scatter(centroids_umap[:, 0], centroids_umap[:, 1], s=30, color='black', label='Centroid')
    
    plt.legend(fontsize=15, frameon=True, edgecolor='black')
    #plt.legend().set_visible(False)
    plt.grid(False)
    plt.show()

# Elbow Method
def elbow_visualizer(model, k_values, data):
    visualizer = KElbowVisualizer(model, k=k_values, timings=False)
    visualizer.fit(data)        # Fit the data to the visualizer
    visualizer.show()           # Finalize and render the figure
```

```python
# UMAP 2D Projection
umap_reduction(X, relation_col, "UMAP - Data Visualization", n_components=2)
```

```python
umap_reduction(X, cause_col, "UMAP - Data Visualization", n_components=2)
```

K-Means

```python
# Elbow Method 

# Instantiate the clustering model
k_values = range(1, 11)
kmeans = KMeans(init='k-means++', n_init=10, random_state=seed)
# Pass the model and data to the visualizer
elbow_visualizer(kmeans, k_values, X)
```

```python
# Silhouette Score Method

k_values = range(2, 6)

fig, ax = plt.subplots(2, 2, figsize=(15,8))
for k in k_values:
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=seed)
    q, mod = divmod(k, 2)
    
    visualizer = SilhouetteVisualizer(kmeans, colors='yellowbrick', ax=ax[q-1][mod])
    visualizer.fit(X)
```

```python
# Apply KMeans
n_clusters = 3

kmeans = KMeans(n_clusters=n_clusters, init='k-means++', n_init=10, random_state=seed)
kmeans.fit(X)
uns_labels = kmeans.labels_

print("\nLabels: ", np.unique(uns_labels))
#print(uns_labels)

centroids = kmeans.cluster_centers_
```
