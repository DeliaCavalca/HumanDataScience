```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import scipy.stats as stats
```

```python
# Data
labels = ['Women', 'Men']
colors=['#F6C8D3', '#ACD9E9']

homicides_tot = 314
homicides = [111, 203]
homicides_family = [96, 55]
homicides_family_partner = [59, 10]
```

```python
# TOT Homicides Victims by Gender

women_pct = homicides[0] / homicides_tot * 100
men_pct = homicides[1] / homicides_tot * 100

fig, ax = plt.subplots(figsize=(10, 3.4))
plt.title('Homicides Victims by Gender (2024)', pad=55, fontsize=22)

ax.legend(
    handles=[
        plt.Line2D([0], [0], color=colors[0], lw=18, label=f'Women ({women_pct:.1f}%)'),
        plt.Line2D([0], [0], color=colors[1], lw=18, label=f'Men ({men_pct:.1f}%)')
    ],
    loc='upper center',
    bbox_to_anchor=(0.5, 1.32),
    ncol=2,
    frameon=False,
    fontsize=20
)

ax.barh(0, women_pct, color=colors[0])
ax.barh(0, men_pct, left=women_pct, color=colors[1])

ax.set_yticks([])
ax.set_xticks([])
ax.set_xlim(0, 100)

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# % Homicides Women by Family Members
# (rispetto al tot di omicidi di donne)

# Dati
family = homicides_family[0]
non_family = homicides[0] - family
percent = family / homicides[0] * 100


colors_family = [colors[0], '#FAE0E7']

# Grafico a ciambella
fig, ax = plt.subplots(facecolor='white')

wedges, texts = ax.pie(
    [family, non_family],
    colors=colors_family,
    radius=1.75,
    startangle=90,
    wedgeprops=dict(width=0.3, edgecolor='black', linewidth=1.5, facecolor='none'),  # solo bordo
)

# Riempie con colori
for i, w in enumerate(wedges):
    w.set_edgecolor(colors_family[i])
    w.set_facecolor(colors_family[i])

# Testo al centro
ax.text(0, 0, f'{percent:.1f}%', ha='center', va='center', fontsize=45, color=colors_family[0])

# Titolo e layout
#plt.title('Women Homicides in Family Context (2024)', pad=60, fontsize=22)
plt.tight_layout()
plt.show()
```
